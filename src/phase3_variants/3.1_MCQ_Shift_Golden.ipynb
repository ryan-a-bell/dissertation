# Shifting the Golden Answer

```python
# Since we have 4 total answers, if we shift it 0,1,2,3 we have all the combos.
```

```python
import pandas as pd

# Load the dataset from the provided URL
url = "https://huggingface.co/datasets/rabell/SysEngBench/resolve/main/test.csv"
df = pd.read_csv(url)

# Function to shift both choices and the correct answer
def shift_choices_and_answer(row, shift_by=1):
    choices = ['choiceA', 'choiceB', 'choiceC', 'choiceD']
    choice_texts = [row['choiceA'], row['choiceB'], row['choiceC'], row['choiceD']]

    # Rotate the choices and correct answer by the shift amount
    shifted_choices = choice_texts[-shift_by:] + choice_texts[:-shift_by]
    
    # Update the label to reflect the new position of the answer
    correct_index = choices.index(f'choice{row["answer"]}')  # Index of the correct answer
    new_label = (correct_index + shift_by) % 4  # Update label index
    
    return {
        'choiceA': shifted_choices[0],
        'choiceB': shifted_choices[1],
        'choiceC': shifted_choices[2],
        'choiceD': shifted_choices[3],
        'answer': choices[new_label][-1].upper(),  # Get the new answer letter (A, B, C, D)
        'label': new_label  # New label based on shifted position
    }

# Function to create datasets with shifted choices and answers
def create_shifted_datasets_with_labels(df):
    datasets = {}
    for shift in range(1, 5):  # Shift by 1, 2, 3, 4 positions
        df_shifted = df.copy()
        shifted_data = df_shifted.apply(lambda row: shift_choices_and_answer(row, shift), axis=1)
        df_shifted[['choiceA', 'choiceB', 'choiceC', 'choiceD', 'answer', 'label']] = pd.DataFrame(shifted_data.tolist())
        datasets[f"shift_{shift}"] = df_shifted
    
    return datasets

# Create shifted datasets with updated choices and labels
shifted_datasets_with_labels = create_shifted_datasets_with_labels(df)

# Save the updated datasets to CSV files
for shift, df_shifted in shifted_datasets_with_labels.items():
    output_file = f"C:/data/{shift}_shifted_dataset_with_labels.csv"
    df_shifted.to_csv(output_file, index=False)

output_paths_with_labels = {shift: f"C:/data/{shift}_shifted_dataset_with_labels.csv" for shift in shifted_datasets_with_labels}
```
