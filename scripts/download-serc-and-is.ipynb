```python
# Modular approach to API Crossref

import requests
import pandas as pd

def search_crossref(search_terms, issn=None, container_title=None, max_results=40):
    """
    Query Crossref for works matching the search terms in a given journal or venue.

    Parameters
    ----------
    search_terms : str
        Text to search for (e.g., 'artificial intelligence OR language model').
    issn : str, optional
        ISSN of the journal to search.  If provided, uses /journals/{ISSN}/works.
    container_title : str, optional
        Name of the journal/venue.  If provided, uses /works with a filter.
    max_results : int
        Maximum number of results to return.

    Returns
    -------
    pandas.DataFrame
        DataFrame containing Title, Year, DOI and URL for each matching work.
    """
    if issn:
        # Search within a specific journal by ISSN
        endpoint = f"https://api.crossref.org/journals/{issn}/works"
        params = {
            "query": search_terms,
            "select": "DOI,title,issued,URL",
            "sort": "issued",
            "order": "desc",
            "rows": max_results
        }
    elif container_title:
        # Search by container title (journal or conference name)
        endpoint = "https://api.crossref.org/works"
        params = {
            "filter": f"container-title:{container_title}",
            "query": search_terms,
            "select": "DOI,title,issued,URL",
            "sort": "issued",
            "order": "desc",
            "rows": max_results
        }
    else:
        raise ValueError("You must specify either an ISSN or a container_title.")

    response = requests.get(endpoint, params=params, timeout=10)
    response.raise_for_status()
    items = response.json()["message"]["items"]

    records = []
    for item in items:
        title = item.get("title", [""])[0]
        year = item.get("issued", {}).get("date-parts", [[None]])[0][0]
        records.append({
            "Title": title,
            "Year": year,
            "DOI": item.get("DOI"),
            "URL": item.get("URL")
        })
    return pd.DataFrame(records)

# Example usage:
# Search the INCOSE International Symposium by ISSN
df1 = search_crossref("large language model OR artificial intelligence", issn="23345837")
# Search the journal "Systems Engineering" by container title
df2 = search_crossref("artificial intelligence", container_title="Systems Engineering")
print(df1.head())
print(df2.head())
```

```python
import time
import requests
import pandas as pd
import networkx as nx
from tqdm import tqdm
from urllib.parse import quote
```

## Cross ref global search

```python
CROSSREF_BASE = "https://api.crossref.org"
MAILTO = "you@example.com"  # strongly recommended for the "polite pool"
UA = "litmap-builder/0.1 (mailto: you@example.com)"

SESSION = requests.Session()
SESSION.headers.update({
    "User-Agent": UA
})

def crossref_works_cursor(query: str,
                         filters: dict | None = None,
                         select: str | None = None,
                         rows: int = 200,
                         max_items: int = 2000,
                         sort: str = "published",
                         order: str = "desc",
                         sleep_s: float = 0.1):
    """
    Global Crossref search via /works with cursor pagination.

    - query: free-text query (supports boolean-ish text, but treat as text search)
    - filters: dict of Crossref filters -> value, joined by commas
    - select: comma-separated fields (keep short; longer selects can slow queries)
    - rows: up to 1000 (Crossref docs); keep moderate for stability
    - max_items: cap total results per query to avoid runaway
    """
    params = {
        "query": query,
        "rows": rows,
        "cursor": "*",
        "sort": sort,
        "order": order,
        "mailto": MAILTO
    }
    if filters:
        params["filter"] = ",".join([f"{k}:{v}" for k, v in filters.items()])
    if select:
        params["select"] = select

    out = []
    cursor = "*"
    fetched = 0

    while True:
        params["cursor"] = cursor
        r = SESSION.get(f"{CROSSREF_BASE}/works", params=params, timeout=30)
        r.raise_for_status()
        msg = r.json()["message"]

        items = msg.get("items", [])
        out.extend(items)
        fetched += len(items)

        # stop conditions
        if len(items) < rows:
            break
        if fetched >= max_items:
            break

        cursor = msg.get("next-cursor")
        if not cursor:
            break

        time.sleep(sleep_s)

    return out[:max_items]
```

## Feed it mindmap questions

```python
MINDMAP_QUESTIONS = [
    "How are large language models used in model-based systems engineering (MBSE) or digital thread workflows?",
    "LLM agents for requirements engineering traceability and verification planning",
    "AI-assisted trade studies and generative design in systems engineering",
    "LLM-based decision support for architecture evaluation and risk management",
]

# A conservative global filter set you can tweak:
GLOBAL_FILTERS = {
    "type": "journal-article",     # try removing if you also want proceedings, preprints, etc.
    # "from-pub-date": "2022-01-01" # optional: time window
}

# Keep select short for speed (Crossref warns long select lists can slow queries). :contentReference[oaicite:4]{index=4}
SELECT_FIELDS = "DOI,title,issued,published,URL,container-title,type,author,is-referenced-by-count"

def normalize_year(item: dict):
    for key in ("published", "issued"):
        dp = item.get(key, {}).get("date-parts")
        if dp and dp[0]:
            return dp[0][0]
    return None

def item_to_row(item: dict):
    return {
        "DOI": item.get("DOI", ""),
        "Title": (item.get("title") or [""])[0],
        "Year": normalize_year(item),
        "URL": item.get("URL", ""),
        "Venue": (item.get("container-title") or [""])[0],
        "Type": item.get("type", ""),
        "CitedByCount": item.get("is-referenced-by-count", None),
        "Authors": "; ".join(
            f"{a.get('family','')}, {a.get('given','')}".strip(", ")
            for a in (item.get("author") or [])
        )
    }

all_rows = []
hit_map = {}  # DOI -> set(question)

for q in tqdm(MINDMAP_QUESTIONS, desc="Crossref queries"):
    items = crossref_works_cursor(
        query=q,
        filters=GLOBAL_FILTERS,
        select=SELECT_FIELDS,
        rows=200,
        max_items=1500,
        sort="published",
        order="desc",
        sleep_s=0.1
    )

    for it in items:
        doi = it.get("DOI")
        if not doi:
            continue
        hit_map.setdefault(doi.lower(), set()).add(q)
        all_rows.append(item_to_row(it))

df_raw = pd.DataFrame(all_rows)

# Deduplicate by DOI; keep the most recent year first
df = (df_raw
      .sort_values(["Year", "CitedByCount"], ascending=[False, False])
      .drop_duplicates(subset=["DOI"])
      .reset_index(drop=True))

df["MatchedQuestions"] = df["DOI"].str.lower().map(lambda d: sorted(hit_map.get(d, [])))
df.head(10)
```

## Pull back citations for litmap edges

```python
def crossref_work(doi: str):
    # Need URL-encoded DOI for the path
    doi_enc = quote(doi, safe="")
    r = SESSION.get(f"{CROSSREF_BASE}/works/{doi_enc}", params={"mailto": MAILTO}, timeout=30)
    r.raise_for_status()
    return r.json()["message"]

def extract_reference_dois(work_msg: dict):
    refs = work_msg.get("reference") or []
    out = []
    for ref in refs:
        # Crossref reference entries sometimes have DOI under 'DOI'
        d = ref.get("DOI")
        if d:
            out.append(d.lower())
    return out

# choose how many top papers you want to hydrate (start with 200–500)
TOP_N = 300

dois = df["DOI"].dropna().head(TOP_N).tolist()

edges = []   # (citing_doi, cited_doi)
missing_ref_counts = 0

for doi in tqdm(dois, desc="Hydrating references"):
    try:
        msg = crossref_work(doi)
        ref_dois = extract_reference_dois(msg)
        if not ref_dois:
            missing_ref_counts += 1
        for rd in ref_dois:
            edges.append((doi.lower(), rd))
        time.sleep(0.1)
    except requests.HTTPError:
        continue

edges_df = pd.DataFrame(edges, columns=["CitingDOI", "CitedDOI"])
print("Hydrated works with no reference DOIs found:", missing_ref_counts)
edges_df.head()
```

## Build the litmap graph

```python
G = nx.DiGraph()

# Add nodes from your main corpus
for _, r in df.iterrows():
    doi = (r["DOI"] or "").lower()
    if not doi:
        continue
    G.add_node(
        doi,
        title=r["Title"],
        year=r["Year"],
        venue=r["Venue"],
        url=r["URL"],
        cited_by=r["CitedByCount"],
        matched_questions=" | ".join(r["MatchedQuestions"] or [])
    )

# Add edges (citing -> cited). Add cited nodes even if outside corpus.
for citing, cited in edges:
    if citing not in G:
        G.add_node(citing)
    if cited not in G:
        G.add_node(cited)
    G.add_edge(citing, cited)

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())

# Useful derived tables
node_table = pd.DataFrame([
    {"DOI": n, **G.nodes[n]}
    for n in G.nodes
])

edge_table = pd.DataFrame([
    {"CitingDOI": u, "CitedDOI": v}
    for u, v in G.edges
])

# Save for Gephi / Cytoscape / Neo4j import
node_table.to_csv("litmap_nodes.csv", index=False)
edge_table.to_csv("litmap_edges.csv", index=False)
nx.write_graphml(G, "litmap.graphml")

print("Wrote: litmap_nodes.csv, litmap_edges.csv, litmap.graphml")
```
